```python
import pandas as pd
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
df = pd.read_csv(r"D:\EXCELR OFFLINE\Python\NLP\IMDB DATASET.csv")

df.head()
```

```python
#Convert Text to Lowercase
df['review'] = df['review'].str.lower()

df['review'].head()
```

```python
#Remove HTML Tags and URLs
df['review'] = df['review'].apply(
    lambda x: re.sub(r'<.*?>|https?://\S+|www\.\S+', '', str(x))
)

df['review'].head()
```

```python
#Remove Punctuation and Special Character
df['review'] = df['review'].apply(
    lambda x: re.sub(r'[^a-zA-Z0-9\s]', '', x)
)

df['review'].head()
```

```python
chat_words = {
    "u": "you",
    "ur": "your",
    "btw": "by the way",
    "gr8": "great",
    "pls": "please"
}

df['review'] = df['review'].apply(
    lambda x: ' '.join(chat_words.get(word, word)
                       for word in x.split())
)

df['review'].head()
```

```python
#Remove Stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

df['review'] = df['review'].apply(
    lambda x: ' '.join(
        word for word in x.split()
        if word not in stop_words
    )
)

df['review'].head()
```

```python
#Tokenization
nltk.download('punkt_tab')

df['tokens'] = df['review'].apply(word_tokenize)

df[['review', 'tokens']].head()
```

```python
stemmer = PorterStemmer()

df['stemmed'] = df['tokens'].apply(
    lambda words: [stemmer.stem(word) for word in words]
)

df[['tokens', 'stemmed']].head()
```

```python
nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

df['lemmatized'] = df['tokens'].apply(
    lambda words: [lemmatizer.lemmatize(word) for word in words]
)

df[['tokens', 'lemmatized']].head()
```
